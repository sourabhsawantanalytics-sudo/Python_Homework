```python
# WAP to find maximum number in given list using for loop.

a = [34,7,90,57,34,89,23,90,45,90,90,89,89]

max_num = a[0]
for num in a:

    if num>max_num:
        max_num=num

print("Maximum number=", max_num)
```

```python
# WAP to find second maximum number in given list using for loop.

a = [34,7,90,57,34,89,23,90,45,90,90,89,89]

a=set(a)
max_num = 0
second_max = 0

for num in a:

    if num > max_num:
        second_max = max_num
        max_num = num

    elif num > second_max:
        second_max = num

print("Second maximum number =", second_max)
    
```

```python
# WAP to find minimum number in given list using for loop.

a = [34,7,90,57,34,89,23,90,45,90,90,89,89]

min_num = a[0]
for num in a:

    if num < min_num:
        min_num = num

print("Minimum number=", min_num)
```
