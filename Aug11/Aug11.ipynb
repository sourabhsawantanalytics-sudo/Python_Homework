# Assignment Oprator

```python
a = 10
a = a + 5
print(a)
```

### Addition

```python
a += 5
print(a)
```

### Substraction

```python
a -= 5
print(a)
```

### Multiplication

```python
a *= 5
print(a)
```

### Division

```python
a /= 5
print(a)
```

### Floor Division

```python
a //= 5
print(a) 
print(type(a))
```

```python
a = 15
a //= 5
print(a)
```

### Modulo

```python
a %= 5
print(a)
```

### Exponential

```python
a **=5
print(a)
```
