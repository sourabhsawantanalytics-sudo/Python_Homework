# Bitwise Operator 
### Bitwise AND (&)

```python
a = 5
b = 3
print(a & b)  # Bitwise AND
```

```python
a = 0b0101
b = 0b0011
print(a & b)  
```

### Bitwise OR (|)

```python
a = 5
b = 3
print(a | b)
```

### Bitwise XOR (^)

```python
a = 5
b = 3
print(a ^ b)
```

### Bitwise NOT (~)

```python
a = 5
print(~a)
```

### Left Shift (<<)

```python
a = 5
print(a << 2)
```

### Right Shift (>>)

```python
a = 20
print(a >> 2)
```
