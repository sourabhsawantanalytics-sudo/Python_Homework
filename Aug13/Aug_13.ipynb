# Slicing



#### Slicing with negative and positive index

```python
name_1 = "Sourabh"
print(name_1[-3:-1])
print(name_1[1:3])
print(name_1[1:-3])
print(name_1[1:3])
print(name_1[1:3])
print(name_1[0:-3])
print(name_1[-4:-1])
```

#### Slicing with list

```python
fruits = ["apple", "banana", "cherry", "mango", "orange"]
print(fruits[-3:-1])
print(fruits[1:3])
print(fruits[-2:-1])
print(fruits[-2: ])
print(fruits[-2: -5])
print(fruits[-5: -2])
```

#### slicing with number

```python
numb = [1,2,3,4,5,6]
print(numb[3:-1])
print(numb[-3:-5])
print(numb[-5:-3])
print(numb[-6 :])
print(numb[ :-1])
print(numb[ 0: ])
print(numb[ 1: ])
```

```python
numb = [1,2,3,4,5,6]
print(numb[0:1])
print(numb[0:2])
print(numb[0:3])
print(numb[0:4])
print(numb[0:5])
print(numb[0:5])
print(numb[0:4])
print(numb[0:3])
print(numb[0:2])
print(numb[0:1])
```
