1) ##### Voting Eligibility Checker :- Take the user's age and citizenship status (yes/no) as input. Print "Eligible to vote" only if age is 18 or above and citizenship is "yes".

```python
age = int(input("Enter your age here: "))
citizenship = (input("Are you citizen of this country? (Yes/No) "))
if (age > 18) and (citizenship == "Yes"):
    print(f"You are {age} years old thats why you are eligible to voting.")

else:
    print(f"You are {age} years old and not eligible for vote.")
```

##### 2.Traffic Light Decision:- Take the color of a traffic light (red, yellow, green) as input. Use if-elif-else to print "Stop", "Slow down", or "Go" accordingly. Handle invalid input with a message like "Invalid color".

```python
color = input("Enter color name: ")

if color == "red":
    print("Stop")

elif color == "yellow":
    print("Slow Down")

elif color == "green":
    print("Go")

else:
    print("Invalid color")
```
