### Create Table Using While Loop

```python
digit = int(input("Enter a number here: "))
count = 1
while count<=10:
    print(f"{digit}*{count} = {digit*count}")
    count+=1
```

```python
digit = 1
while digit <=10:
    i = 1

    while i <=10:
        print(f"{digit}*{i} = {digit * i}")
        i+=1

    print()
    digit+=1
              
```
