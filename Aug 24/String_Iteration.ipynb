### Iterate list & inside string

```python
a=["Jay", "Veeru", "Basanti", "Gabber"]
for i in a:
    print(i)
    for char in i:
        print(char)
```

```python
b =["123456", "987654", "Maharashtra", "Pune"]
for ch in b:
    for d in ch:
        print(d)
```

```python
fruits = ["Apple", "Banana", "Kiwi"]
print(fruits)
print(fruits[::-1])
for fruit in fruits:
    for i in fruit:
        print(i)
```

```python
#
Mobile = ["Samsung", "Apple", "Nokia", "Mi"]
for M in Mobile:
    for c in M:
        print(c, end="---->")
        
```
